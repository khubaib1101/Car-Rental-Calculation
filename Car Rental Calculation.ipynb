```python
# Car Rental System 17
def calculate_rental(car_type,days,driver=False):
  car_type = car_type.lower()
  if car_type =='economy':
    car_payment = 3000
  elif car_type =='sedan':
    car_payment = 5000
  elif car_type == 'suv':
    car_payment = 8000
  else:
    print("Enter proper car type")
  car_cost = car_payment * days
  if days >=7:
    discount = car_cost * 0.15
  elif days>=3 and days <=6:
    discount = car_cost * 0.10
  else:
    discount = 0
  if driver:
    driver_cost =  2000*days
  else:
    driver_cost = 0
  final_bill = car_cost - discount + driver_cost
  print("Car cost--> =",car_cost)
  print("Driver Cost",driver_cost)
  print("Discount",discount)
  print("__________---------___________")
  print("Final bill",final_bill)
calculate_rental('sedan',7,driver=False)
```
